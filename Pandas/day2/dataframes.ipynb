```python
'''
🔍 Real-World Use Case:
Imagine you work for a school's admin panel. You receive Excel sheets with students' names, subjects, marks, and attendance. These files must be cleaned, analyzed, and visualized.
In such cases, a Pandas DataFrame lets you easily manage the data as tables inside Python.
'''

# What is a DataFrame?
# A dataframe is a two-dimentional, size-mutable, and heterogeneous data structure with labeled axes (rows and columns). It's like a digital spreadsheet or SQL table built into Python.
```

```python
import pandas as pd
data = {
    'Name': [
        'Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack',
        'Kira', 'Liam', 'Mona', 'Nina', 'Oscar', 'Paul', 'Quincy', 'Rachel', 'Steve', 'Tina',
        'Uma', 'Victor', 'Wendy', 'Xander', 'Yara', 'Zane', 'Alice', 'Bob', 'Charlie', 'David'
    ],
    'Subject': [
        'Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math',
        'Science', 'History', 'Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math', 'Science',
        'History', 'Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math', 'Science', 'History'
    ],
    'Marks': [
        85, 92, 78, 88, 74, 81, 95, 67, 89, 73,
        91, 76, 84, 88, 79, 90, 66, 93, 87, 69,
        82, 85, 90, 74, 78, 92, 81, 89, 77, 83
    ]
}

df = pd.DataFrame(data)
# print(df)

# | **Term**        | **Definition**                                                                                                                                         |
# | --------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------ |
# | **DataFrame**   | A 2D labeled data structure with columns of potentially different data types. Each column is a `Series`. Think of it like an Excel sheet or SQL table. |
# | **Row**         | A horizontal line of data in the DataFrame, representing one record (like a student).                                                                  |
# | **Column**      | A vertical line of data in the DataFrame, representing a field or attribute (like "Marks" or "Subject").                                               |
# | **Index**       | A label or identifier for each row. By default, it's numeric (0, 1, 2...), but it can also be custom (like student IDs or dates).                      |
# | **.shape**      | Returns a tuple (rows, columns) indicating the size of the DataFrame.                                                                                  |
# | **.columns**    | Shows the names (labels) of all columns.                                                                                                               |
# | **.index**      | Shows the range or labels assigned to each row.                                                                                                        |
# | **.dtypes**     | Tells you the data type of each column (like int64, float64, object for strings).                                                                      |
# | **.head(n)**    | Displays the first `n` rows of the DataFrame (default is 5).                                                                                           |
# | **.tail(n)**    | Displays the last `n` rows of the DataFrame.                                                                                                           |
# | **.info()**     | Provides a summary: number of non-null entries, column names, data types, and memory usage.                                                            |
# | **.describe()** | Returns descriptive statistics for numerical columns — count, mean, std, min, 25%, 50%, 75%, max.                                                      |

print(df.shape) # (Total Rows of Dataframe, Total Columns of Dataframe)
print(df.columns)
print(df.index)
print(df.dtypes)
print(df.head(10))
print(df.tail(10))
print(df.info())
print(df.describe())
```

```python
# Objective: Simulate a small student grade report using Pandas.
# 🛠 Steps:
# Create a DataFrame with 10 rows (students).
# Columns: Name, Subject, Marks, Grade
# Add new column: Pass/Fail based on Marks (>= 40 is Pass)
# Show:
# .describe() for numerical summary
# .info() for structure
# List of only failed students
import pandas as pd
import matplotlib.pyplot as plt
data = {
    'Name': [
        'Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack',
    ],
    'Subject': [
        'Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math', 'Science', 'History', 'Math',
    ],
    'Marks': [
        85, 92, 78, 88, 74, 81, 39, 67, 39, 73,
    ],
    'Grade':['A+','A+','A','A+','A','A+','F','A+','F','A-']
}
student_data = pd.DataFrame(data)
# print(student_data)
student_data['Result'] = student_data['Marks']>=40
student_data['Result'] = student_data['Result'].replace({True:'Pass',False:'Fail'})
print(student_data)

print(student_data.describe())
print(student_data.info())

print(student_data[student_data['Result'] == 'Fail'])
student_data['Result'].value_counts().plot(kind='bar',title='Pass vs Fail')
plt.show()
```
