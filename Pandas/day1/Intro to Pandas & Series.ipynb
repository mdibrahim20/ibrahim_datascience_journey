```python
"""
Imagine you are working for a budget tracking app startup. Users input daily expenses, and you need to analyze and summarize that data — how much they spend daily, what category, and more. But before building full reports, you need to represent their expense data efficiently. That's where Pandas Series comes in — it's the foundation for handling labeled one-dimensional data like temperature readings, stock prices, or attendance logs.
"""

"""
1# What is pandas?
-> A powerful open-source data manipulation library for Python.
-> Designed for structured data: tabular (DataFrames) and labeled 1D data (Series).
-> Built on top of NumpPy for speed

✅ Real Use: Used by data analysts at Netflix to clean and analyze user viewing history.

2# What is a Pandas Series?

A Pandas Series is a one-dimensional labeled array that can hold any data type -- integers,floats,strings etc. You can think of it as a column in an Excel sheet or a single list with labels.

3. Installing and Importing Pandas
-> pip install pandas
-> import pandas as pd
"""

```

```python
## -------------- Pandas Series ----------------

import pandas as pd
import numpy as np
# From a list
marks = pd.Series([11,25,36,47])

# From a dictionary
expenses = pd.Series({'food':200,'transport':45,'books':150});

# from a numpy array
arr = np.array([10,28,55,65])
s = pd.Series(arr)

print(marks)
print(expenses)
print(s)
## -------------- Pandas Series End ----------------

```

```python
## -------------- Indexing & Slicing ----------------
print(marks[0])
print(expenses['food'])
print(marks[1:3])
## -------------- Indexing & Slicing ----------------
```

```python
## -------------- Broadcasting & Vectorization ----------------
# Vectorization: It means performing operations on entire array or series without using loops.
# Broadcasting: How pandas/numpy automatically expands scalar values or smaller arrays to match the shape of larager arrays/Series during operations.

marks + 5  # It adds 5 to each data
expenses * 2
## -------------- Broadcasting & Vectorization ----------------
```

```python
## -------------- Comparison Operations ----------------

marks>70
expenses>150
## -------------- Comparison Operations ----------------
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Weekly expense data (actual values)
weekly_expenses = pd.Series({
    'day1': 33,
    'day2': 32,
    'day3': 34,
    'day4': 31,
    'day5': 32,
    'day6': 35,
    'day7': 31.45
})

# Apply 18% tax (final cost after tax)
weekly_expenses_with_tax = weekly_expenses * 1.18

print("✅ Weekly Expenses With 18% Tax:\n", weekly_expenses_with_tax)
print("📊 Total Weekly Expense (with tax):", round(weekly_expenses_with_tax.sum(), 2))
print("📈 Average Daily Expense (with tax):", round(weekly_expenses_with_tax.mean(), 2))

# Visualize the data
plt.figure(figsize=(8, 5))
weekly_expenses_with_tax.plot(kind='bar', color='skyblue', title='Weekly Expenses with 18% Tax')
plt.ylabel("Amount in tk")
plt.xlabel("Days")
plt.grid(True)
plt.tight_layout()
plt.show()
```
